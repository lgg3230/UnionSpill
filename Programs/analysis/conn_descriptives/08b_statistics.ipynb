pUrpose od this code is to generate the following statistics: 

1) Conditional on having posive flows, how many share same industry-microregion cell?

2) amonfg the top 5, 10, 20 and 25% of connectivity measure (among those pairs with positive connectivity), how many share the same industry and microregion

```python
# import libraries

import argparse
import math
import duckdb
import pandas as pd
```

```python
# read dataset from parquet:

df = pd.read_parquet('/kellogg/proj/lgg3230/UnionSpill/Data/RAIS_aux/bilateral_regression_data.parquet')
```

```python
df_sample = df.sample(n=100000, random_state=42)
```

```python
# Restrict to rows that have positive pretreatment flows
df_sample = df_sample[df_sample['has_flow_pre'] ==1 ] 

```

```python
df_positve = df[df['has_flow_pre']==1]
```

```python
# out of the positve flows, what proportion share the same industry -microregion cell?
f_positve =  len(df_positve[df_positve['same_industry_micro']==1]) / len(df_positve)
print(f_positve)
```

```python
# out of the  top 5, 10, 20 and 25% of pairs with the highest pretreatment flows, what proportion share the same industry-microregion cell?
flow_thresholds = [0.05, 0.10, 0.20, 0.25]
for threshold in flow_thresholds:
    flow_cutoff = df_positve['bilateral_conn_pre'].quantile(1 - threshold)
    df_top = df_positve[df_positve['bilateral_conn_pre'] >= flow_cutoff]
    f_top = len(df_top[df_top['same_industry_micro'] == 1]) / len(df_top)
    print(f"Top {int(threshold*100)}%: {f_top}")
```
